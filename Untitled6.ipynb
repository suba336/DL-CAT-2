```python
import pandas as pd

file_path = '/content/extracted_data/Indian Rainfall Dataset District-wise Daily Measurements.csv'

# Read dataset with correct delimiter
df = pd.read_csv(file_path, sep=';')

# Strip quotes from column names
df.columns = [col.replace('"', '').strip() for col in df.columns]

# Clean string columns
for col in ['state', 'district']:
    if col in df.columns:
        df[col] = df[col].astype(str).str.replace('"', '').str.strip()

print("Cleaned columns:", df.columns.tolist())
print("\nTotal rows in dataset:", len(df))

# Show sample districts
unique_districts = df[['state', 'district']].drop_duplicates().reset_index(drop=True)
print(f"\nNumber of unique districts: {len(unique_districts)}")
display(unique_districts.head(15))
```

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Select an active district from the list (e.g., East Khasi Hills in Meghalaya)
selected_district = "East Khasi Hills"
district_df = df[df['district'] == selected_district].copy()

# Helper to get the correct column name with ordinal suffixes
def get_col_suffix(day_num):
    if 11 <= day_num <= 13:
        return f"{day_num}th"
    elif day_num % 10 == 1:
        return f"{day_num}st"
    elif day_num % 10 == 2:
        return f"{day_num}nd"
    elif day_num % 10 == 3:
        return f"{day_num}rd"
    else:
        return f"{day_num}th"

day_cols = [get_col_suffix(i) for i in range(1, 32)]

# Melt from wide to long format
long_df = district_df.melt(
    id_vars=['state', 'district', 'month'],
    value_vars=day_cols,
    var_name='day_str',
    value_name='rainfall'
)

# Extract numeric day
long_df['day'] = long_df['day_str'].str.extract(r'(\d+)').astype(int)
long_df['year'] = 2022

# Filter out invalid calendar dates
def make_date_or_nan(row):
    try:
        return pd.Timestamp(year=int(row['year']), month=int(row['month']), day=int(row['day']))
    except ValueError:
        return pd.NaT

long_df['date'] = long_df.apply(make_date_or_nan, axis=1)
long_df = long_df.dropna(subset=['date'])

# Sort chronologically
time_series = long_df.sort_values('date').reset_index(drop=True)[['date', 'rainfall']]

# Handle missing values: coerce errors and interpolate
time_series['rainfall'] = pd.to_numeric(time_series['rainfall'], errors='coerce')
time_series['rainfall'] = time_series['rainfall'].interpolate(method='linear').fillna(0)

# Normalize rainfall measurements using MinMaxScaler
scaler = MinMaxScaler(feature_range=(0, 1))
time_series['scaled_rainfall'] = scaler.fit_transform(time_series[['rainfall']])

print(f"Processed time-series for {selected_district}. Shape: {time_series.shape}")
display(time_series.head(10))
```

### Sequential Learning Model Selection & Justification

For rainfall prediction, we select an **LSTM (Long Short-Term Memory)** network. LSTMs are specialized Recurrent Neural Networks (RNNs) that excel at processing sequential data because they use gating mechanisms (input, forget, and output gates) to selectively retain or discard information over time. This makes them highly effective for weather forecasting where short-to-medium-term patterns (like a multi-day storm system or consecutive dry days) have strong temporal dependencies that traditional feedforward neural networks or standard RNNs (which suffer from vanishing gradients) cannot capture easily.

```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from sklearn.metrics import mean_absolute_error, mean_squared_error

# 1. Generate input-output sequences using the previous 7 days
def create_sequences(data, seq_length=7):
    X, y = [], []
    for i in range(len(data) - seq_length):
        X.append(data[i:(i + seq_length)])
        y.append(data[i + seq_length])
    return np.array(X), np.array(y)

sequence_length = 7
scaled_data = time_series['scaled_rainfall'].values

X, y = create_sequences(scaled_data, sequence_length)

# Reshape X to be [samples, time steps, features]
X = X.reshape((X.shape[0], X.shape[1], 1))

# 2. Chronological split (80% training, 20% testing)
split_index = int(len(X) * 0.8)

X_train, X_test = X[:split_index], X[split_index:]
y_train, y_test = y[:split_index], y[split_index:]

# Keep track of dates for plotting
test_dates = time_series['date'].iloc[sequence_length + split_index:].reset_index(drop=True)

print(f"Sequence dataset created for {selected_district}.")
print(f"Train size: {X_train.shape}, Test size: {X_test.shape}")
```

```python
# 3. Define and train the LSTM model
model = Sequential([
    LSTM(50, activation='relu', input_shape=(sequence_length, 1), return_sequences=True),
    Dropout(0.2),
    LSTM(50, activation='relu'),
    Dropout(0.2),
    Dense(1)
])

model.compile(optimizer='adam', loss='mse')

print("Training the LSTM model on new district data...")
history = model.fit(
    X_train, y_train,
    epochs=50,
    batch_size=16,
    validation_split=0.1,
    verbose=1
)
```

```python
# 4. Evaluate the model and make predictions
predictions_scaled = model.predict(X_test)

# Inverse transform to original scale (rainfall in mm)
predictions = scaler.inverse_transform(predictions_scaled).flatten()
y_test_original = scaler.inverse_transform(y_test.reshape(-1, 1)).flatten()

# Calculate evaluation metrics
mae = mean_absolute_error(y_test_original, predictions)
rmse = np.sqrt(mean_squared_error(y_test_original, predictions))

print(f"\nEvaluation Metrics for {selected_district}:")
print(f"Mean Absolute Error (MAE): {mae:.4f} mm")
print(f"Root Mean Squared Error (RMSE): {rmse:.4f} mm")
```

```python
# 5. Plot actual versus predicted rainfall
plt.figure(figsize=(14, 6))
plt.plot(test_dates, y_test_original, label='Actual Rainfall', color='blue', marker='o')
plt.plot(test_dates, predictions, label='Predicted Rainfall', color='orange', linestyle='--', marker='x')
plt.title(f'Actual vs Predicted Daily Rainfall for {selected_district} (2022)')
plt.xlabel('Date')
plt.ylabel('Rainfall (mm)')
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Demonstrate predictions for some sample unseen dates (last 5 days of the test set)
demo_df = pd.DataFrame({
    'Date': test_dates[-5:].values,
    'Actual Rainfall (mm)': y_test_original[-5:],
    'Predicted Rainfall (mm)': predictions[-5:]
})
print("\nDemonstration of predictions for the final 5 days of the testing sequence:")
display(demo_df)
```

```python
import pandas as pd

# Calculate which districts have the most active/varying rainfall to select a better candidate
# Melt a subset to calculate standard deviation of daily measurements

print("Searching for districts with the highest rainfall activity/variation in 2022...")

temp_long = df.melt(
    id_vars=['state', 'district'],
    value_vars=day_cols,
    value_name='rainfall'
)
temp_long['rainfall'] = pd.to_numeric(temp_long['rainfall'], errors='coerce').fillna(0)

# Group and find variation
district_stats = temp_long.groupby(['state', 'district'])['rainfall'].agg(['mean', 'max', 'std']).reset_index()
active_districts = district_stats.sort_values(by='std', ascending=False).head(15)

print("\nTop 15 districts with highest rainfall variation (best for sequential prediction):")
display(active_districts)
```
